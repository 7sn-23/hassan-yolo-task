```python
import os
import cv2
import glob

INPUT_IMAGES_DIR = '/content/drive/MyDrive/blender dataset/images'
INPUT_LABELS_DIR = '/content/drive/MyDrive/blender dataset/labels'

OUTPUT_IMAGES_DIR = '/content/drive/MyDrive/blender dataset/images_cropped'
OUTPUT_LABELS_DIR = '/content/drive/MyDrive/blender dataset/labels_cropped'

os.makedirs(OUTPUT_IMAGES_DIR, exist_ok=True)
os.makedirs(OUTPUT_LABELS_DIR, exist_ok=True)

CROP_XMIN = 700
CROP_XMAX = 1200
CROP_YMIN = 350
CROP_YMAX = 850

image_files = glob.glob(os.path.join(INPUT_IMAGES_DIR, '*.jpg')) + glob.glob(os.path.join(INPUT_IMAGES_DIR, '*.png'))

for img_path in image_files:
    img_name = os.path.basename(img_path)
    base_name, ext = os.path.splitext(img_name)

    img = cv2.imread(img_path)
    if img is None:
        continue

    h, w, _ = img.shape

    ymin = max(0, CROP_YMIN)
    ymax = min(h, CROP_YMAX)
    xmin = max(0, CROP_XMIN)
    xmax = min(w, CROP_XMAX)

    cropped_img = img[ymin:ymax, xmin:xmax]

    new_img_name = f"{base_name}_crop{ext}"
    cv2.imwrite(os.path.join(OUTPUT_IMAGES_DIR, new_img_name), cropped_img)

    lbl_path = os.path.join(INPUT_LABELS_DIR, f"{base_name}.txt")
    if os.path.exists(lbl_path):
        with open(lbl_path, 'r') as f:
            lines = f.readlines()

        new_lines = []
        for line in lines:
            parts = line.strip().split()
            if len(parts) == 5:
                cls_id, x_c, y_c, box_w, box_h = map(float, parts)

                orig_xmin = (x_c - box_w / 2) * w
                orig_ymin = (y_c - box_h / 2) * h
                orig_xmax = (x_c + box_w / 2) * w
                orig_ymax = (y_c + box_h / 2) * h

                new_xmin = max(0, orig_xmin - xmin)
                new_ymin = max(0, orig_ymin - ymin)
                new_xmax = min(xmax - xmin, orig_xmax - xmin)
                new_ymax = min(ymax - ymin, orig_ymax - ymin)

                new_w = xmax - xmin
                new_h = ymax - ymin

                if new_w > 0 and new_h > 0:
                    new_xc = ((new_xmin + new_xmax) / 2) / new_w
                    new_yc = ((new_ymin + new_ymax) / 2) / new_h
                    new_bw = (new_xmax - new_xmin) / new_w
                    new_bh = (new_ymax - new_ymin) / new_h

                    if 0 <= new_xc <= 1 and 0 <= new_yc <= 1:
                        new_lines.append(f"{int(cls_id)} {new_xc} {new_yc} {new_bw} {new_bh}\n")

        new_lbl_name = f"{base_name}_crop.txt"
        with open(os.path.join(OUTPUT_LABELS_DIR, new_lbl_name), 'w') as f:
            f.writelines(new_lines)

print("Cropping and label adjustment complete!")
```

```python
import os
import cv2
import glob
import random
import math
import numpy as np

def adjust_brightness_contrast(img, brightness=0, contrast=0):
    if brightness != 0:
        if brightness > 0:
            shadow = brightness
            highlight = 255
        else:
            shadow = 0
            highlight = 255 + brightness
        alpha = (highlight - shadow) / 255.0
        gamma = shadow
        img = cv2.addWeighted(img, alpha, img, 0, gamma)
    if contrast != 0:
        f = 131 * (contrast + 127) / (127 * (131 - contrast))
        alpha = f
        gamma = 127 * (1 - alpha)
        img = cv2.addWeighted(img, alpha, img, 0, gamma)
    return img

def generate_augmented_dataset(image_dir, labels_dir, output_img_dir, output_lbl_dir, target_size=(640, 640), target_count=250):
    os.makedirs(output_img_dir, exist_ok=True)
    os.makedirs(output_lbl_dir, exist_ok=True)

    image_files = glob.glob(os.path.join(image_dir, '*.jpg')) + glob.glob(os.path.join(image_dir, '*.png'))
    if not image_files:
        print("No images found in the specified cropped directory!")
        return

    augmentations_per_image = max(1, math.ceil(target_count / len(image_files)))

    for img_path in image_files:
        img_name = os.path.basename(img_path)
        base_name, ext = os.path.splitext(img_name)
        lbl_path = os.path.join(labels_dir, base_name + '.txt')

        img = cv2.imread(img_path)
        if img is None:
            continue

        bboxes = []
        if os.path.exists(lbl_path):
            with open(lbl_path, 'r') as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) == 5:
                        bboxes.append([int(parts[0]), float(parts[1]), float(parts[2]), float(parts[3]), float(parts[4])])

        # Base Image
        resized_base = cv2.resize(img, target_size)
        cv2.imwrite(os.path.join(output_img_dir, f"{base_name}_base.jpg"), resized_base)
        if bboxes:
            with open(os.path.join(output_lbl_dir, f"{base_name}_base.txt"), 'w') as f:
                for box in bboxes:
                    f.write(f"{int(box[0])} {box[1]} {box[2]} {box[3]} {box[4]}\n")

        # Augmented Variations
        for i in range(augmentations_per_image):
            aug_img = adjust_brightness_contrast(img, brightness=random.randint(-35, 35), contrast=random.randint(-15, 15))
            if i % 2 == 0:
                aug_img = cv2.GaussianBlur(aug_img, (5, 5), 0)
            aug_resized = cv2.resize(aug_img, target_size)

            cv2.imwrite(os.path.join(output_img_dir, f"{base_name}_photo_{i}.jpg"), aug_resized)
            if bboxes:
                with open(os.path.join(output_lbl_dir, f"{base_name}_photo_{i}.txt"), 'w') as f:
                    for box in bboxes:
                        f.write(f"{int(box[0])} {box[1]} {box[2]} {box[3]} {box[4]}\n")

            if i % 2 == 0:
                flipped_img = cv2.flip(img, 1)
                flipped_resized = cv2.resize(flipped_img, target_size)
                cv2.imwrite(os.path.join(output_img_dir, f"{base_name}_geom_flip_{i}.jpg"), flipped_resized)
                if bboxes:
                    with open(os.path.join(output_lbl_dir, f"{base_name}_geom_flip_{i}.txt"), 'w') as f:
                        for box in bboxes:
                            cls, x_c, y_c, bw, bh = box
                            new_x_c = 1.0 - x_c
                            f.write(f"{int(cls)} {new_x_c} {y_c} {bw} {bh}\n")
            else:
                rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
                rotated_resized = cv2.resize(rotated_img, target_size)
                cv2.imwrite(os.path.join(output_img_dir, f"{base_name}_geom_rot_{i}.jpg"), rotated_resized)
                if bboxes:
                    with open(os.path.join(output_lbl_dir, f"{base_name}_geom_rot_{i}.txt"), 'w') as f:
                        for box in bboxes:
                            cls, x_c, y_c, bw, bh = box
                            new_x_c = y_c
                            new_y_c = 1.0 - x_c
                            f.write(f"{int(cls)} {new_x_c} {new_y_c} {bh} {bw}\n")

    total_images = len(glob.glob(os.path.join(output_img_dir, '*.jpg'))) + len(glob.glob(os.path.join(output_img_dir, '*.png')))
    total_labels = len(glob.glob(os.path.join(output_lbl_dir, '*.txt')))
    print(f"Augmentation Done! Generated {total_images} images and {total_labels} labels in augmented folders.")

if __name__ == '__main__':
    input_images = '/content/drive/MyDrive/blender dataset/images_cropped'
    input_labels = '/content/drive/MyDrive/blender dataset/labels_cropped'

    augmented_dir = '/content/augmented_output_images'
    augmented_labels_dir = '/content/augmented_output_labels'

    generate_augmented_dataset(input_images, input_labels, augmented_dir, augmented_labels_dir)
```

```python
import os
import glob
import shutil
import random
import yaml

from google.colab import drive
drive.mount('/content/drive')

def split_and_save_to_drive(augmented_img_dir, augmented_lbl_dir, final_dataset_dir):
    splits = ['train', 'val', 'test']
    for split in splits:
        os.makedirs(os.path.join(final_dataset_dir, 'images', split), exist_ok=True)
        os.makedirs(os.path.join(final_dataset_dir, 'labels', split), exist_ok=True)

    all_files = glob.glob(os.path.join(augmented_img_dir, '*.jpg')) + glob.glob(os.path.join(augmented_img_dir, '*.png'))
    base_names = list(set([os.path.splitext(os.path.basename(f))[0].split('_base')[0].split('_photo')[0].split('_geom')[0]
                           for f in all_files]))

    random.shuffle(base_names)

    train_split = int(len(base_names) * 0.7)
    val_split = int(len(base_names) * 0.9)

    train_bases = base_names[:train_split]
    val_bases = base_names[train_split:val_split]
    test_bases = base_names[val_split:]

    def move_files(bases, split_name):
        for base in bases:
            matched_imgs = glob.glob(os.path.join(augmented_img_dir, f"{base}*"))
            for img_path in matched_imgs:
                img_name = os.path.basename(img_path)
                lbl_name = os.path.splitext(img_name)[0] + '.txt'
                lbl_path = os.path.join(augmented_lbl_dir, lbl_name)

                shutil.copy(img_path, os.path.join(final_dataset_dir, 'images', split_name, img_name))
                if os.path.exists(lbl_path):
                    shutil.copy(lbl_path, os.path.join(final_dataset_dir, 'labels', split_name, lbl_name))

    move_files(train_bases, 'train')
    move_files(val_bases, 'val')
    move_files(test_bases, 'test')

    data_config = {
        'path': final_dataset_dir,
        'train': 'images/train',
        'val': 'images/val',
        'test': 'images/test',
        'nc': 1,
        'names': ['car']
    }

    yaml_path = os.path.join(final_dataset_dir, 'data.yaml')
    with open(yaml_path, 'w') as f:
        yaml.dump(data_config, f, sort_keys=False)

    print(f"Dataset successfully split and saved to: {final_dataset_dir}")

if __name__ == '__main__':
    augmented_dir = '/content/augmented_output_images'
    augmented_labels_dir = '/content/augmented_output_labels'

    final_dataset_local = '/content/final_yolo_dataset'
    final_dataset_drive = '/content/drive/MyDrive/final_yolo_dataset'

    split_and_save_to_drive(augmented_dir, augmented_labels_dir, final_dataset_local)
    split_and_save_to_drive(augmented_dir, augmented_labels_dir, final_dataset_drive)
```

```python
def draw_yolo_boxes(image_dir, labels_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)
    for img_path in glob.glob(os.path.join(image_dir, '*.jpg')) + glob.glob(os.path.join(image_dir, '*.png')):
        img_name = os.path.basename(img_path)
        base_name = os.path.splitext(img_name)[0]
        lbl_path = os.path.join(labels_dir, base_name + '.txt')

        img = cv2.imread(img_path)
        if img is None:
            continue
        h, w, _ = img.shape

        if os.path.exists(lbl_path):
            with open(lbl_path, 'r') as f:
                lines = f.readlines()
            for line in lines:
                parts = line.strip().split()
                if len(parts) == 5:
                    cls_id, x_c, y_c, box_w, box_h = map(float, parts)
                    xmin = int((x_c - box_w / 2) * w)
                    ymin = int((y_c - box_h / 2) * h)
                    xmax = int((x_c + box_w / 2) * w)
                    ymax = int((y_c + box_h / 2) * h)
                    cv2.rectangle(img, (xmin, ymin), (xmax, ymax), (0, 255, 0), 2)

        cv2.imwrite(os.path.join(output_dir, img_name), img)
    print("Boxes drawn and saved successfully!")

if __name__ == '__main__':
    draw_yolo_boxes('/content/augmented_output_images', '/content/augmented_output_labels', '/content/visualized_output')
```
